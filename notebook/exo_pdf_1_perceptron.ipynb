# Exercice 1 : Perceptrion simple [ Porte AND ]



```python
import numpy as np
import pandas as pd
```

```python
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1]) # PORTE AND
```

```python
df_x = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
print("Dataframe X :")
display(df_x)
```

```python
# Creéation d'un Dataframe pour y

df_y = pd.DataFrame(y, columns=['Target'])
print("\nDataFrame y : ")
display(df_y)
```

```python
# Eape 3 - Initialiser les poids et le biais

np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()
print("Poids initiaux : ", w)
print("Biais initial : ", b)
```

```python
# Etape 4 - Fonction d'activation (sigmoide)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
# Etape 5 - Définir la fonction du perceptron

def predict(X, w, b):
    z = np.dot(X, w) + b
    return sigmoid(z)
```

```python
# Etape 6 - D€finir la fonction de perte (MSE)

def loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
# Etape 7 - Entrainement du perceptron (descente de gradient)

learning_rate = 0.1
epochs = 2000

for i in range(epochs):
    y_pred = predict(X, w, b)
    # Calcul du gradient
    dw = -2 * np.dot((y - y_pred) * y_pred * (1 - y_pred), X) / len(X)
    db = -2 * np.mean((y - y_pred) * y_pred * (1 - y_pred))

    # Mise à jour des poids
    w -= learning_rate * dw
    b -= learning_rate * db

    if i % 100 == 0:
        print(f"Epoch {i}, loss: {loss(y, y_pred)}")
```

```python
# Epate 8 - Tester le perceptron

y_pred = predict(X, w, b)
print("Prédiction finales : ", y_pred)
print("Classes prédites (seuil 0.5): ", (y_pred > 0.5).astype(int))
```

```python
# Bonus : Entropie croisée

np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()

def bce(y_true, y_pred, eps=1e-12):
    return -np.mean(y_true * np.log(y_pred + eps) + (1 - y_true) * np.log(1 - y_pred + eps))

learning_rate = 0.5

for i in range(2000):
    y_pred = predict(X, w, b)
    dz = y_pred - y # Gradient de l'entropie croisée
    w -= learning_rate * np.dot(dz, X) / len(X)
    b -= learning_rate * np.mean(dz)
    if i % 400 == 0:
        print(f"Epoch {i}, Loss: {bce(y, y_pred):.4f}")

print("Prédictions finales : ", predict(X, w, b).round(3))
print("Classes prédites (seuil 0.5): ", (y_pred > 0.5).astype(int))
```
