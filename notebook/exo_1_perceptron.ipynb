## Exercice 1 : Le Perceptron

Objectif : resoudre une classification binaire (porte logique AND) avec un neurone et la descente de gradient.

### Porte AND

```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1]) # porte logique AND

np.random.seed(42) # aleatoire mais pas trop
w = np.random.randn(2) # poids
b = np.random.randn() # biais

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict(X, w, b):
    return sigmoid(X @ w + b)

def bce(y, p, eps=1e-12): # entropie croisée binaire
    return - np.mean(y * np.log(p + eps) + (1 - y) * np.log(1 - p + eps))

lr, epochs = 0.5, 2000 # learning rate et epoque (~iteration)
for i in range(epochs):
    p = predict(X, w, b)
    dz = p - y
    w -= lr * (X.T @ dz) / len(X) # X.T (transposition de la matrice)
    b -= lr * np.mean(dz)
    if i % 200 == 0:
        print(f"Epoch {i} : Loss = {bce(y, p):.4f}")

print("Predictions :", predict(X, w, b).round(3))
```
